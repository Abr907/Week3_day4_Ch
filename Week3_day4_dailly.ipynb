```python
#Q1.Compute the circle’s area.
import math

class Circle:
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2


# Test
circle1 = Circle(5)

print("Radius:", circle1.radius)
print("Area:", circle1.area())
```

```python
# Q3. Add two circles together and return a new circle with the new radius

class Circle:
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2

    def __str__(self):
        return f"Circle with radius {self.radius}"

    def __add__(self, other):
        return Circle(self.radius + other.radius)


circle1 = Circle(5)
circle2 = Circle(3)

circle3 = circle1 + circle2

print(circle3)
print("New radius:", circle3.radius)
print("New area:", circle3.area())
```

```python
#Q6. Store multiple circles in a list and sort them — implement
import math

class Circle:
    def __init__(self, radius):
        if radius <= 0:
            raise ValueError("Radius must be positive.")

        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2

    def __str__(self):
        return f"Circle with radius {self.radius}"

    def __add__(self, other):
        if not isinstance(other, Circle):
            return NotImplemented

        return Circle(self.radius + other.radius)

    def __gt__(self, other):
        if not isinstance(other, Circle):
            return NotImplemented

        return self.radius > other.radius

    def __eq__(self, other):
        if not isinstance(other, Circle):
            return NotImplemented

        return self.radius == other.radius

    def __lt__(self, other):
        if not isinstance(other, Circle):
            return NotImplemented

        return self.radius < other.radius


# Create circles
circle1 = Circle(7)
circle2 = Circle(2)
circle3 = Circle(5)
circle4 = Circle(3)

# Store them in a list
circles = [circle1, circle2, circle3, circle4]

# Sort from smallest to largest
circles.sort()

print("Circles sorted by radius:")

for circle in circles:
    print(circle, "| Area:", round(circle.area(), 2))
```

```python
import math
import matplotlib.pyplot as plt


class Circle:
    def __init__(self, radius):
        if radius <= 0:
            raise ValueError("Radius must be positive.")
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2

    def __str__(self):
        return f"Circle(radius={self.radius})"

    def __add__(self, other):
        if not isinstance(other, Circle):
            return NotImplemented
        return Circle(self.radius + other.radius)

    def __gt__(self, other):
        if not isinstance(other, Circle):
            return NotImplemented
        return self.radius > other.radius

    def __eq__(self, other):
        if not isinstance(other, Circle):
            return NotImplemented
        return self.radius == other.radius

    def __lt__(self, other):
        if not isinstance(other, Circle):
            return NotImplemented
        return self.radius < other.radius


# Create circles
circles = [
    Circle(30),
    Circle(10),
    Circle(25),
    Circle(15),
    Circle(20)
]

# Sort from smallest to largest
circles.sort()

# Set up the matplotlib plot
fig, ax = plt.subplots(figsize=(10, 5))
ax.set_aspect('equal')

# Draw each circle horizontally
x_position = 0
spacing = 80

for circle in circles:
    radius = circle.radius

    # Create a circle patch centered at (x_position, radius)
    # so that they all sit on the bottom horizontal line (y=0)
    draw_circle = plt.Circle((x_position, radius), radius, fill=False, edgecolor='blue', linewidth=2)
    ax.add_patch(draw_circle)

    # Write the radius below the circle
    ax.text(x_position, -10, f"Radius: {radius}", ha='center', va='top', fontsize=10)

    x_position += spacing

# Configure limits and remove axes ticks for a clean look
ax.set_xlim(-50, x_position - spacing + 50)
ax.set_ylim(-30, 80)
ax.axis('off')
plt.title("Sorted Circles - Python OOP")
plt.show()
```
